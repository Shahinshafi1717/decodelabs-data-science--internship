```python
import pandas as pd
import numpy as np

print(" Libraries imported successfully!")
```

```python
titanic = pd.read_csv('https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv')
print(" Dataset loaded successfully!")
print("First look at data:")
print(titanic.head())
```

```python
print("Dataset Size:", titanic.shape)
print("\nColumn Names:", titanic.columns.tolist())
print("\nData Types:")
print(titanic.dtypes)
print("\nBasic Info:")
titanic.info()
```

```python
print("Missing Values in Each Column:")
print(titanic.isnull().sum())
print("\nMissing Value Percentage:")
print((titanic.isnull().sum() / len(titanic)) * 100)
```

```python
titanic['Age'] = titanic['Age'].fillna(titanic['Age'].median())
titanic['Embarked'] = titanic['Embarked'].fillna(titanic['Embarked'].mode()[0])
if 'Cabin' in titanic.columns:
    titanic.drop(columns=['Cabin'], inplace=True)
    print("Cabin column dropped!")
else:
    print("Cabin already removed!")
print("\nMissing values after fixing:")
print(titanic.isnull().sum())
print("\n All missing values handled!")
```

```python
print("Duplicates before:", titanic.duplicated().sum())
titanic.drop_duplicates(inplace=True)
print("Duplicates after:", titanic.duplicated().sum())
print(" Duplicate check done!")
```

```python
titanic['Sex'] = titanic['Sex'].map({'male': 0, 'female': 1})
titanic['Embarked'] = titanic['Embarked'].map({'S': 0, 'C': 1, 'Q': 2})
print("✅ Text converted to numbers!")
print(titanic[['Sex', 'Embarked']].head(10))
```

```python
titanic.drop(columns=['Name', 'Ticket', 'PassengerId'], inplace=True)
print("Unnecessary columns removed!")
print("Remaining columns:", titanic.columns.tolist())
print("Final shape:", titanic.shape)
print("\nFinal cleaned data:")
print(titanic.head())
```

```python
titanic.to_csv('cleaned_titanic.csv', index=False)
print("Cleaned dataset saved!")
print("\nSummary of what I did:")
print("1. Loaded dataset - 891 rows, 12 columns")
print("2. Found missing values - Age, Cabin, Embarked")
print("3. Fixed missing values - Filled Age & Embarked, Dropped Cabin")
print("4. Checked duplicates - None found")
print("5. Converted text to numbers - Sex and Embarked")
print("6. Removed useless columns - Name, Ticket, PassengerId")
print("7. Final dataset - 891 rows, 8 columns")
print("\nTASK 2 COMPLETED!")
```

My Conclusions from Data Cleaning:

1. Cabin column had too many missing values (77%) so I dropped it
2. Age was filled with median to avoid effect of outliers
3. Embarked had only 2 missing values so mode worked fine
4. Converting text to numbers is important for ML models
5. Clean data is the foundation of good analysis
