```python
import pandas as pd
import numpy as np

print("Libraries imported successfully!")
```

```python
titanic = pd.read_csv('https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv')

print("Dataset loaded successfully!")
print("\nFirst look at the data:")
print(titanic.head())
```

```python
print("Dataset Size:")
print(f"Total Rows (Passengers): {titanic.shape[0]}")
print(f"Total Columns (Features): {titanic.shape[1]}")
print("\nColumn Names:")
for col in titanic.columns:
    print(f"  - {col}")
```

```python
print("Data Types of Each Column:")
print(titanic.dtypes)

print("\nWhat each column means:")
print("  PassengerId - Unique ID for each passenger")
print("  Survived    - 0 = Not Survived, 1 = Survived")
print("  Pclass      - Ticket class (1=1st, 2=2nd, 3=3rd)")
print("  Name        - Name of passenger")
print("  Sex         - Gender of passenger")
print("  Age         - Age of passenger")
print("  SibSp       - Number of siblings/spouses onboard")
print("  Parch       - Number of parents/children onboard")
print("  Ticket      - Ticket number")
print("  Fare        - Passenger fare price")
print("  Cabin       - Cabin number")
print("  Embarked    - Port of embarkation (S/C/Q)")
```

```python
print("Statistical Summary:")
print(titanic.describe())
print("\nCategorical Columns Summary:")
print(titanic.describe(include='object'))
```

```python
print("Missing Values in Dataset:")
print(titanic.isnull().sum())
print("\nMissing Value Percentage:")
print((titanic.isnull().sum() / len(titanic)) * 100)
```

```python
print("=" * 50)
print("MY DATASET UNDERSTANDING REPORT - Shahin")
print("=" * 50)

print("\nAbout the Dataset:")
print("  Name: Titanic Passenger Dataset")
print("  Source: Kaggle")
print("  This dataset contains information about")
print("  passengers who were on the Titanic ship in 1912")

print("\nDataset Size:")
print(f"  Total Passengers: {titanic.shape[0]}")
print(f"  Total Features: {titanic.shape[1]}")

print("\nData Types:")
print("  Numerical columns: PassengerId, Survived, Pclass, Age, SibSp, Parch, Fare")
print("  Categorical columns: Name, Sex, Ticket, Cabin, Embarked")

print("\nMissing Values Found:")
print("  Age     - 19.86% missing")
print("  Cabin   - 77.10% missing")
print("  Embarked - 0.22% missing")

print("\nInteresting Facts:")
print(f"  Most passengers were male: {titanic['Sex'].value_counts()['male']} males")
print(f"  Most boarded from Southampton: {titanic['Embarked'].value_counts()['S']} passengers")
print(f"  Average age: {titanic['Age'].mean():.1f} years")
print(f"  Average fare: ${titanic['Fare'].mean():.2f}")

print("\nWhat this data represents:")
print("  This dataset helps us understand who survived")
print("  the Titanic disaster based on their age,")
print("  gender, class and other features")

print("\nTASK 1 COMPLETED SUCCESSFULLY! ")
```
