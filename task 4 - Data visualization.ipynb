```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_style("whitegrid")
plt.rcParams['figure.figsize'] = (10, 6)
print("Libraries imported successfully!")
```

```python
titanic = pd.read_csv('https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv')
titanic['Age'] = titanic['Age'].fillna(titanic['Age'].median())
titanic['Embarked'] = titanic['Embarked'].fillna(titanic['Embarked'].mode()[0])
print("Dataset loaded and ready for visualization!")
print("Shape:", titanic.shape)
```

```python
plt.figure(figsize=(8, 6))
colors = ['#ff6b6b', '#51cf66']
titanic['Survived'].value_counts().plot(kind='bar', color=colors, edgecolor='black')
plt.title('Chart 1: Survival Count of Titanic Passengers', fontsize=14)
plt.xlabel('0 = Not Survived  |  1 = Survived', fontsize=12)
plt.ylabel('Number of Passengers', fontsize=12)
plt.xticks(rotation=0)
for i, v in enumerate(titanic['Survived'].value_counts()):
    plt.text(i, v + 5, str(v), ha='center', fontsize=12, fontweight='bold')
plt.tight_layout()
plt.show()
print("Chart 1 created!")
print("Insight: More passengers did NOT survive (549) than survived (342)")
```

```python
plt.figure(figsize=(8, 6))
colors = ['#74c0fc', '#f783ac']
titanic.groupby('Sex')['Survived'].mean().mul(100).plot(
    kind='bar', color=colors, edgecolor='black')
plt.title('Chart 2: Survival Rate by Gender', fontsize=14)
plt.xlabel('Gender', fontsize=12)
plt.ylabel('Survival Rate (%)', fontsize=12)
plt.xticks(rotation=0)
for i, v in enumerate(titanic.groupby('Sex')['Survived'].mean().mul(100)):
    plt.text(i, v + 1, f'{v:.1f}%', ha='center', fontsize=12, fontweight='bold')
plt.tight_layout()
plt.show()
print("Chart 2 created!")
print("Insight: Female survival rate (74.2%) is much higher than male (18.9%)")
```

```python
plt.figure(figsize=(8, 6))
colors = ['#ffd43b', '#c0c0c0', '#cd7f32']
titanic.groupby('Pclass')['Survived'].mean().mul(100).plot(
    kind='bar', color=colors, edgecolor='black')
plt.title('Chart 3: Survival Rate by Passenger Class', fontsize=14)
plt.xlabel('Passenger Class (1=First, 2=Second, 3=Third)', fontsize=12)
plt.ylabel('Survival Rate (%)', fontsize=12)
plt.xticks(rotation=0)
for i, v in enumerate(titanic.groupby('Pclass')['Survived'].mean().mul(100)):
    plt.text(i, v + 1, f'{v:.1f}%', ha='center', fontsize=12, fontweight='bold')
plt.tight_layout()
plt.show()
print("Chart 3 created!")
print("Insight: 1st class had highest survival (63%) vs 3rd class (24%)")
```

```python
plt.figure(figsize=(10, 6))
plt.hist(titanic[titanic['Survived']==1]['Age'],
         bins=30, alpha=0.7, color='#51cf66', label='Survived', edgecolor='black')
plt.hist(titanic[titanic['Survived']==0]['Age'],
         bins=30, alpha=0.7, color='#ff6b6b', label='Not Survived', edgecolor='black')
plt.title('Chart 4: Age Distribution by Survival', fontsize=14)
plt.xlabel('Age', fontsize=12)
plt.ylabel('Number of Passengers', fontsize=12)
plt.legend(fontsize=12)
plt.tight_layout()
plt.show()
print("Chart 4 created!")
print("Insight: Children (age 0-10) had higher survival rate than adults")
```

```python
plt.figure(figsize=(10, 6))
plt.hist(titanic[titanic['Survived']==1]['Fare'],
         bins=30, alpha=0.7, color='#51cf66', label='Survived', edgecolor='black')
plt.hist(titanic[titanic['Survived']==0]['Fare'],
         bins=30, alpha=0.7, color='#ff6b6b', label='Not Survived', edgecolor='black')
plt.title('Chart 5: Fare Distribution by Survival', fontsize=14)
plt.xlabel('Fare Price ($)', fontsize=12)
plt.ylabel('Number of Passengers', fontsize=12)
plt.legend(fontsize=12)
plt.tight_layout()
plt.show()
print("Chart 5 created!")
print("Insight: Passengers who paid higher fare survived more")
```

```python
plt.figure(figsize=(10, 8))
numerical_cols = titanic[['Survived', 'Pclass', 'Age', 'SibSp', 'Parch', 'Fare']]
sns.heatmap(numerical_cols.corr(),
            annot=True,
            fmt='.2f',
            cmap='coolwarm',
            linewidths=0.5)
plt.title('Chart 6: Correlation Between Features', fontsize=14)
plt.tight_layout()
plt.show()
print("Chart 6 created!")
print("Insight: Pclass has strong negative correlation with Survival")
print("Insight: Fare has positive correlation with Survival")
```
